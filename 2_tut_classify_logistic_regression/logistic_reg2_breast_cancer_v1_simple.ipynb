
# Logistic Regression

In this project, we will see how to train a logistic regression model. This is intended as an introduction to logistic regression. 

We will be working with the Breast Cancer dataset, which contains some very detailed measurements of cells. Along with each observation of measurements, we have the diagnosis of the cell (malignant or not). Our goal is to train a model that will be able to predict whether or not a given cell is malignant given only its measurements.

Steps:

```
--> Collect data 
--> Split into training + testing
--> Feed training data to model (i.e. train the model)
--> Evaluate the model by feeding testing data
```


````python
import numpy as np 
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt

# import warnings
# warnings.filterwarnings('ignore')
````

## Collect data 

````python
df = pd.read_csv("breast_cancer.csv")
df.head()

# print(df['radius_mean']) # this way I can access each colum
````

````python
print(df.info())
````

````python
print(df.describe())
````

````python
# Label Encoding: Turn target variable into 1s and 0s

df['diagnosis'] = df['diagnosis'].map({'M':1,'B':0})
print(df.head(4))
````

The first thing to do is to do is to separate the target variable (here called "y") and the predictors (here called "X"). Note that we use an uppercase X as convention in order to mymic the mathematical language. 

In mathematics, an uppercase symbol represents that the variable is multidimensional (a matrix).

````python
# drop id and empty column

df.drop(['Unnamed: 32'], axis=1, inplace=True)
````

````python
# check missing values

print(df.isnull().sum())
````

````python
# Prepare the data feature X and label y
# X = df(['radius_mean','texture_mean', ...and so on]) # takes a long time

X = df.drop(["diagnosis"], axis=1) # our predictors
y = df["diagnosis"] # our target variable
````

# Split the dataset into training / testing

````python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)
````

````python
print(X_train.shape)
print(X_test.shape)
````

````python
print(f'X_train: {X_train[:5]}')
````

## Feed training data to the model (i.e. train the model)

````python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()

# Train the model on the training data
model.fit(X_train, y_train)

# Observ: Here data is not scaled, so will receive warning
````

````python
# I took first sample from test data and its actual label was 1
X_sample = [[842302.0, 17.99, 10.38, 122.8, 1001.0, 0.1184, 0.2776, 0.3001, 0.1471, 0.2419, 0.07871, 1.095, 0.9053, \
           8.589, 153.4, 0.006399, 0.04904, 0.05373, 0.01587, 0.03003, 0.006193, 25.38, 17.33, 184.6, 2019.0, 0.1622,\
           0.6656, 0.7119, 0.2654, 0.4601, 0.1189],]

X_df = pd.DataFrame(X_sample, columns=X_train.columns)

y_pred = model.predict(X_df)

print("predicted:", y_pred)
````

##  Evaluate the model by feeding the testing data

````python
## Predict the target variable on the test data and also its actual value
y_pred = model.predict(X_test)

print("predicted:", y_pred)
print("actual   :", list(y_test))
````

````python
from sklearn.metrics import accuracy_score

# Evaluate the accuracy of the model
accuracy = accuracy_score(y_test, y_pred)
print(f'Accuracy: {accuracy:.2f}')
````
