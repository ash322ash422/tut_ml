columns and their meaning:

- total_bill: Total bill (in USD) for the meal, including food and drinks
- tip: Tip amount given by the customer
- sex: Gender of the person paying the bill — "Male" or "Female"
- smoker: Whether the customer is a smoker — "Yes" or "No"
- day: Day of the week — "Thur", "Fri", "Sat", or "Sun"
- time: Time of the meal — "Lunch" or "Dinner"
- size: Number of people at the table (party size)

```python
# Lets visualize above value counts with pie plot
#'autopct' displays the percentage upto the 1 decimal
#'radiius' sets the radius of pie plot
plt.figure(figsize = (10, 5))
data   = df['day'].value_counts()
labels = df['day'].value_counts().index.tolist() # IMPORTANT: the labels should match with the numbers in data
print(data)
print(labels)

plt.pie(data, autopct="%.1f%%", radius=1, labels=labels)

plt.show()
```

```python
#I want some exotic exploded pie chart

plt.figure(figsize = (10, 5))
data   = df['day'].value_counts()
labels = df['day'].value_counts().index.tolist()
print(data)
print(labels)

plt.pie(data, autopct="%.1f%%", radius=1, labels=labels, explode=[0,0,0.2,0.4])

plt.show()
```

```python
#donut pie plot using little trickery

plt.figure(figsize = (10, 5))
data   = df['day'].value_counts()
labels = df['day'].value_counts().index.tolist()
print(data)
print(labels)

plt.pie(data, autopct="%.1f%%", radius=1, labels=labels)

circle = plt.Circle((0,0), 0.5, color='white')
plot = plt.gcf()
plot.gca().add_artist(circle)

plt.show()
```

```python
# lets look at distribution of total_bill
x = df['total_bill'] # get the column
# print(x)
sns.histplot(x,
             kde=True, # This plots the smooth curve
             color="skyblue"
)
plt.title("Histogram with KDE of Total Bill")
plt.show()
```

## observation:
* Most people in this dataset spent between $10–$20 on their meal.
* A few people spent a lot more (up to ~$50), which the KDE curve will show as a long tail to the right. It is **right skewed**

```python
# Plot KDE for each day
sns.histplot(data=df, 
             x='total_bill', 
             kde=True, 
             hue='day', 
             multiple='dodge', 
             color="skyblue")
plt.title("Histogram with KDE of Total Bill by Day")
plt.show()
```

```python
# Now I want to see above seperately:

days = ['Thur', 'Fri', 'Sat', 'Sun']

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle('KDE of Total Bill by Day')
for day, ax in zip(days, axes.flat):
    sns.kdeplot(data=df[df['day'] == day], 
                x='total_bill', 
                fill=True,
                ax=ax,
                color='skyblue')
    ax.set_title(day)
    ax.set_xlabel('Total Bill')
    
plt.tight_layout()
plt.show()

# I do not like this one.
```

```python
# STEP1 : draw box plot of a column
plt.boxplot(df['total_bill'])

plt.title('Boxplot Of Total Bill Amount', fontsize = 18)
plt.ylabel('Total Bill', fontsize = 15)

plt.show()
```

```python
# STEP2: I want to print the Q1, Q2, Q3, IQR, outliers etc
# Lets do box plot of individual columns:
# Inter - Quantile Range (IQR - For Total Bill)
Q1 = df['total_bill'].quantile(0.25)
Q2 = df['total_bill'].quantile(0.50)
Q3 = df['total_bill'].quantile(0.75)
print(f"Q1: {Q1}")
print(f"Q2: {Q2}")
print(f"Q3: {Q3}")


IQR = Q3 - Q1
print(IQR)

# Define the bounds for outliers
lower_bound = Q1 - 1.5 * IQR # min
upper_bound = Q3 + 1.5 * IQR # max
print(f"lower bound: {lower_bound}")
print(f"upper bound: {upper_bound}")

# Find the outliers and filter the DataFrame to remove outliers
df_outliers = df[(df['total_bill'] < lower_bound) | (df['total_bill'] > upper_bound)]

# Display the outliers
print(f"Outlier Data Points:\n{df_outliers}")
```

## observation:
### Diagonal Plots: These are histograms (or KDEs) of each individual numeric variable, split by sex.

- total_bill diagonal plot: Shows the distribution of total bills (Males may have a slightly wider range, suggesting they might pay for higher bills more often.)
- tip diagonal plot: Distribution of tip values. (Both sexes give tips in similar ranges, though you spot that male group tends to give slightly more.)
- size diagonal plot: Shows group sizes. (Most group sizes are 2–4, regardless of gender.)

### total_bill vs tip:

- Positive correlation: higher total bill → higher tip (as expected).

### total_bill vs size:
- Generally, larger parties lead to larger bills.
- Points cluster around party size 2–4.
- Sex coloring can reveal if, say, males tend to be in slightly larger groups.
