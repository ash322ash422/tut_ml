# From Scikit-learn

````python
import sklearn.datasets as ds
````

````python
# step1: view all data files available

all_datasets = []

for name in dir(ds):
    if name.startswith("load_") or name.startswith("fetch_"):
        all_datasets.append(name)


for d in all_datasets:
    print(d)
````

### Commonly used load_ datasets (small / toy datasets)
- These come bundled with sklearn (no internet required):
```
load_diabetes
load_digits
load_files
load_iris
```

### fetch_ datasets (larger / real-world)

These download data from the internet:
```
fetch_20newsgroups
fetch_20newsgroups_vectorized
fetch_california_housing
```


# View dataset details (structure)

````python
from sklearn.datasets import load_iris

iris = load_iris()

print(iris.keys())
````

#### Lets look at **data, target, target_names, DESCR, feature_names**

````python
print(iris.DESCR)
````

````python
print(iris.data[:10])
````

````python
print(iris.feature_names)
````

````python
print(iris.target[:10])
````

````python
print(iris.target_names)
````

# Save it in CSV file

````python
# STEP1: Put the data in dataframe

df = pd.DataFrame(iris.data, columns=iris.feature_names)

print(df.sample(10))
````

````python
df["target"] = iris.target

print(df.sample(10))
````

````python
# Add class names instead of numbers:
df["species"] = df["target"].apply(lambda x: iris.target_names[x])

print(df.sample(10))
````

````python
# 2) save it in CSV file
df.to_csv("junk_iris1.csv", index=False) # I do not want index
````
