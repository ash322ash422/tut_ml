# Sample Run of All regressor models on diabetes dataset 
- X = patient data (age, BMI, blood pressure, etc.)
- y = a numeric score representing disease progression after one year

```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.utils import all_estimators

# X, y = load_diabetes(return_X_y=True)
data = load_diabetes()
X, y = data.data, data.target


X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

for name, Model in all_estimators(type_filter='regressor'):
    try:
        model = Model()
        model.fit(X_train, y_train)
        score = model.score(X_test, y_test)
        print(f"{name}: {score:.3f}")
    except Exception as e:
        pass
```

```python
# If you  want to see the column names and target values
print("X column names:", data.feature_names)
print("y:", y)
```

# Sample Run of All classifier models on iris dataset 

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.utils import all_estimators

# X, y = load_iris(return_X_y=True)
data = load_iris()
X, y = data.data, data.target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

for name, Model in all_estimators(type_filter='classifier'):
    try:
        model = Model()
        model.fit(X_train, y_train)
        score = model.score(X_test, y_test)
        print(f"{name}: {score:.3f}")
    except:
        pass
```

```python
# If you  want to see the column names and target values
print("X column names:", data.feature_names)
print("y:", y)
```
