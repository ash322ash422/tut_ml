# Fix data types

```python
# I want to see the info: Include deep memory usage (actual memory for object dtypes)

print(df.info(memory_usage="deep"))

# In the beginning, pandas will use big memory consuming data types: 
# int64, float64, object

# observ: int64 implies that numbers are integers
```

```python
# Lets look at descriptive stats so we know the min and max value of our data: i.e. range
# This allows us to chhose the right data type

print(df.describe())

# observ: mpg ranges from min=9 to max=46.6 And so on.
```

```python
# for MPG, the range of values are from min=9.0 to max=46.6. 
# So we change the datatype from float64 -> float16
df['mpg'] = df['mpg'].astype('float16')
```

```python
# I want to see the info: Include deep memory usage (actual memory for object dtypes)

print(df.info(memory_usage="deep"))

# obser: There is small drop in memory consumption
```

```python
# Now we focus on cylinders: It seems that this has finite numbers and is categorical
# step1: value counts of cylinder
print(df['cylinders'].value_counts())
```

```python
# step2: For cylinders, the range of values are from min=3 to max=8 
# So the datatype change from int64 -> int8. 
# Do not change this to category, especially if you want to use ML model later.
df['cylinders'] = df['cylinders'].astype('int8')
```

```python
# I want to see the info: Include deep memory usage (actual memory for object dtypes)

print(df.info(memory_usage="deep"))

# obser: There is small drop in memory consumption
```

```python
# For displacement, the range of values are from min=68 to max=455. All values are whole numbers
df['displacement'] = df['displacement'].astype('int16')
```

```python
# I want to see the info: Include deep memory usage (actual memory for object dtypes)

print(df.info(memory_usage="deep"))

# obser: There is small drop in memory consumption
```

```python
# For horsepower, the range of values are from min=46 to max=230. All values are whole numbers
df['horsepower'] = df['horsepower'].astype('int16')
```

```python
# I want to see the info: Include deep memory usage (actual memory for object dtypes)

print(df.info(memory_usage="deep"))

# obser: There is small drop in memory consumption
```

# outliers

```python
# Detect outliers in the displacement column using the IQR method
import pandas as pd
col = "weight" # horsepower , weight,displacement

df = pd.read_csv('data_MPG.csv', encoding='ascii')

q1 = df[col].quantile(0.25)
q3 = df[col].quantile(0.75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

print(lower_bound)
print(upper_bound)
print(outliers[[col]].head())
print(len(outliers))

# obser: horsepower has outliers
# weight and displacement has no outliers
```
