# Why **data types matter**

### **1. Memory efficiency**

Different data types consume very different amounts of RAM.

* `int64` → 8 bytes
* `int32` → 4 bytes
* `float64` → 8 bytes
* `category` → often **<1 byte per value**

**Impact:**
Large datasets can go from **GBs → MBs** just by choosing correct dtypes.

---

### **2. Computational speed**

CPUs and vectorized libraries (NumPy, pandas, PyTorch) are optimized for **native numeric types**.

* Numeric dtypes → fast, vectorized operations
* Wrong dtype → Python-level loops, slow execution

**Impact:**
Same algorithm, different dtype → **2×–10× speed difference**

---

### **3. Model correctness & behavior**

ML models expect specific dtypes.

* Trees treat `int` differently from `category`
* Linear models break if strings sneak in
* Neural networks require `float32/float16`


---

> **If your data types are wrong, your memory is wasted.
> If memory is wasted, your model will never survive production.**
