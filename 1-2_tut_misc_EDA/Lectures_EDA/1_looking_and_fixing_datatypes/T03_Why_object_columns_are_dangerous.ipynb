# Why **object columns are dangerous**
### **1. Massive hidden memory usage**

`object` columns store **Python objects**, not raw values.

**Impact:**
An `object` column can use **5–10× more memory** than `category`.

---

### **2. Extremely slow operations**

Operations on `object` columns:

* Cannot be vectorized efficiently
* Often run in Python loops

**Impact:**
EDA and feature engineering become **painfully slow** on large datasets.

---

### **3. ML incompatibility**

Most ML algorithms **do not understand `object` dtype**.
