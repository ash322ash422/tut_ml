Here I load this dataset into a database

```python
import pandas as pd
import sqlite3

filename = "ecommerce_dirty_dataset_5000.csv"
```

```python
# first find column names from csv
# Prompt used: How to read column names from a csv file.
import csv

with open(filename, newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    headers = reader.fieldnames

print(headers)
```

```python
# Load CSV direclty in DB
df = pd.read_csv(filename)
```

```python
# OPTIONAL:TODO Here change the data type in dataframe using appropriate methods
```

```python
# Connect to SQLite (DB file will be created if it doesn't exist)
conn = sqlite3.connect("commerce.db")

# Write to SQLite (table auto-created)
df.to_sql(
    name="employees",       # table name
    con=conn,
    if_exists="replace",    # 'fail', 'replace', or 'append'
    index=False             # don't create index column
)

conn.close()

print("CSV loaded into SQLite successfully!")
```
