# Clean up and change data types

```python
# Do the necessary import
import pandas as pd
import numpy as np

import seaborn as sns
import matplotlib.pyplot as plt
file_name = "HR_Analytics_Data.csv"
```

```python
# Step1: Reading the wrong way. This file contains records that start with quotes "
df = pd.read_csv(file_name)
print(df.sample(6))
```

```python
# Step2 : Read the file as plain text
with open(file_name, 'r', encoding='utf-8') as f:
    lines = f.read().splitlines()

# Replace literal '\t' with actual tab, and strip quotes
cleaned_lines = [line.strip('"').replace('\\t', '\t') for line in lines]

# Write to a temporary string buffer for pandas to read
from io import StringIO
buffer = StringIO("\n".join(cleaned_lines))

# Read the cleaned data
df = pd.read_csv(buffer, sep='\t')
```

### Now lets save memory: Memory optimization

```python
# Lets see columns types:
print(df.dtypes)
# Age values they range from 18 to 60. For this column you do not need int64
# Same logic to other columns
```

```python
# Lets print memory usage before memory optimization

print(df.memory_usage(deep=True))
print(f"Before optimization: {df.memory_usage(deep=True).sum() / 1024:.2f} KB")
```

```python
# Lets drop columns that we do not need: "EmployeeCount", "EmployeeNumber", etc and other columns if needed

df.drop(columns=["EmployeeCount", "EmployeeNumber", "StandardHours"], inplace=True)
# Side Note: Here after dropping columns, we do not need to reindex. But after dropping rows, reindexing is recommended

print(df.memory_usage(deep=True))
print(f"Memory now: {df.memory_usage(deep=True).sum() / 1024:.2f} KB")
```

```python
# Downcast integer int64 columns to int8, int16, int32, int64 based on the data range present
int_cols = df.select_dtypes(include=['int64']).columns
df[int_cols] = df[int_cols].apply(pd.to_numeric, downcast='integer')

print(df.memory_usage(deep=True))
print(f"Memory now: {df.memory_usage(deep=True).sum() / 1024:.2f} KB")
```

```python
# Lets verify that downcast worked
print(df.dtypes)
```

```python
# Convert object to Categorical Text Columns
obj_cols = df.select_dtypes(include=['object']).columns

for col in obj_cols:
    num_unique = df[col].nunique()
    num_total = len(df[col])
    if num_unique / num_total < 0.5:  # heuristic: if < 50% unique
        print(f"Converting {col} to datatype category")
        df[col] = df[col].astype('category')


print(df.memory_usage(deep=True))
print(f"Memory now: {df.memory_usage(deep=True).sum() / 1024:.2f} KB")
```

```python
# Lets verify that downcast worked
print(df.dtypes)
```

```python
# Now lets convert columns that contain binary values (like yes, no or y, n) to Boolean

#step1: identify those columns
bool_cols = [col for col in df.columns 
             if df[col].dropna().astype(str).str.lower().isin(['yes', 'no', 'y', 'n']).all()]

print(bool_cols)
```

```python
# Lets print few values from these columns:
print(df[bool_cols].sample(10))
```

```python
# Lets confirm that these columns hold binary values:
print(df[['Attrition', 'Over18', 'OverTime']].nunique()) 

# The column over18 contains only 1 unique value: Y. This can be dropped
```

```python
# Drop column Over18
df.drop(columns=["Over18",], inplace=True)
```

```python
# step2: Normalize: strip empty space, convert all values to lower case and then map them to True/False

bool_cols = ['Attrition', 'OverTime']

mappings = {'yes': True,
            'no': False,
}

df[bool_cols] = df[bool_cols].apply(
    lambda x: x.astype(str).str.strip().str.lower().map(mappings)
)
```

```python
# step3: Convert to boolean dtype to save memory
df[bool_cols] = df[bool_cols].astype('bool')

print(df.memory_usage(deep=True))
print(f"Memory now: {df.memory_usage(deep=True).sum() / 1024:.2f} KB")

# You would see that Attrition columns memory usage went down from 1697 B to 1470 B
```
