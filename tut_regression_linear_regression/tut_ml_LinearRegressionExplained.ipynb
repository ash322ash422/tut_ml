# Linear Regression 1
The simple `LinearRegression` estimator of 1 variable is used to find equation of line:
$$
y = a_0 + a_1 x_1
$$
Geometrically, this is akin to fitting a plane to points in 2 dimensions.
Here $x_1$ is a feature. Here **y is dependant variable and x1 is an indep var.**


**QUESTION:** If you had studied longer, would your overall scores get any better?

One way of answering this question is by having data on how long you studied for and what scores you got. We can then try to see if there is a pattern in that data, and if in that pattern, when you add to the hours, it also ends up adding to the scores percentage.

For instance, say you have an hour-score dataset, which contains entries such as (1.5 hr, 87.5% score),  (1.61 hr, 78% score), etc . The kind of data type that can have any intermediate value (or any level of 'granularity') is known as continuous data

Here we are trying to find equation of line, 
$$
y = a_0 + a_1 x_1 
$$

that can describe relation between $x_1$ (hours) and y (scores). $a_0$ is the intercept/bias and $a_1$ is the $x_1$-coefficient or the weight of $x_1$.


```python
import pandas as pd
import numpy as np

# Lets import some visual lib.
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
plt.style.use('dark_background')
```

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([5, 30, 20, 40, 60])

df = pd.DataFrame({"X": x, "Y": y})
print(df.to_string(index=False))
```

```python
# Train Model
X = df[['X']].values
y = df['Y'].values
model = LinearRegression()
model.fit(X, y)

# Equation
print("Intercept/bias (a0):", round(model.intercept_,3))
print("Slope/weight (a1)  :", round(model.coef_[0],3))

# Prediction
print("Predicted score for 6 hours:", round(model.predict([[6]])[0],3))

# Plot
# Set the x and y axis limits and ticks
plt.xlim(0, 6) # Set x-axis limits from 0 to 6
plt.ylim(0, 70) # Set y-axis limits from 0 to 70
plt.yticks(np.arange(0, 71, 10)) # Set y-axis ticks with a step of 10

plt.scatter(df['X'], df['Y'], color='gold')
plt.plot(df['X'], model.predict(X), color='yellow')
plt.xlabel('Hours Studied')
plt.ylabel('Scores Obtained')
plt.title('Hours vs Scores (Linear Regression)')
plt.show()
```

```python

# Data points
x = np.array([1, 2, 3, 4, 5])
y = np.array([5, 30, 20, 40, 60])

df = pd.DataFrame({"X": x, "Y": y})
print(df.to_string(index=False))
```

# Line 1

```python
# Calculate y values for the line at given x values
line_eq = "y = 5 + 10x"
color = "gold"
y_line = 10 * x + 5  # Compute predicted y values using y = 10x + 5

# Create the plot
# plt.figure(figsize=(6, 4))

# Scatter plot of actual data points
plt.scatter(x, y, color='gold', label='Actual Data Points', marker='o')

# Scatter plot for y = 10x + 5 at given x values
plt.scatter(x, y_line, color=color, label=f'Line Points ({line_eq})', marker='x')

# Connect the red points to form a line
plt.plot(x, y_line, color=color)

# Annotate each point
for i in range(len(x)):
    plt.text(x[i], y[i], f"({x[i]},{y[i]})", fontsize=10, verticalalignment='bottom', horizontalalignment='right')
    plt.text(x[i], y_line[i], f"({x[i]},{y_line[i]})", fontsize=10, verticalalignment='top', horizontalalignment='left', color=color)

# Set the x and y axis limits and ticks
plt.xlim(0, 6) # Set x-axis limits from 0 to 6
plt.ylim(0, 70) # Set y-axis limits from 0 to 70
plt.yticks(np.arange(0, 71, 10)) # Set y-axis ticks with a step of 10

# Labels, title, and legend
plt.xlabel('Hours Studied')
plt.ylabel('Scores Obtained')
plt.title('Hours vs Scores (Linear Regression)')
plt.legend()
# plt.grid(True)

# Show the plot
plt.show()

# Print table of original and predicted values
print("Original Data Points:")
print("\n".join(f"({x[i]}, {y[i]})" for i in range(len(x))))

print(f"\nLine Points ({line_eq}):")
print("\n".join(f"({x[i]}, {y_line[i]})" for i in range(len(x))))

# Compute RSS error terms
errors = (y - y_line) ** 2
RSS = np.sum(errors)  # Residual Sum of Squares

# Print error table
print("\nTable of Errors:")
print(f"{'X':<5}{'Actual Y':<10}{'Predicted Y':<15}{'Error Term (Squared)':<20}")
print("-" * 50)
for i in range(len(x)):
    print(f"{x[i]:<5}{y[i]:<10}{y_line[i]:<15}{errors[i]:<20.2f}")

# Print total RSS
print(f"\nResidual Sum of Squares (RSS): {RSS:.2f}")
```

# Explanation of above
The above error terms are calulcated as follows:

error1 = (actual_y1 - predicted_y1)^2 = (5-15)^2 = (10)^2 = 100

error2 = (actual_y2 - predicted_y2)^2 = (30-25)^2 = (5)^2 = 25  (and so on)

error3 = (actual_y3 - predicted_y3)^2 = 225

error4 = (actual_y4 - predicted_y4)^2 = 25

error5 = (actual_y5 - predicted_y5)^2 = 25

RSS = error1 + error2 + error3 + error4 + error5 = 100 + 25 + 225 + 25 + 25 = 400.

# Goal of linear regression
- In above case, between line1, line2 and line3, the RSS for line3 is minimum.
- Find the line ( y = b + mx  ) that minimizes the RSS. Finding the line means determining the slope(m) and y-intercept (b).
- Linear Regression attempts to find m and b.



# Lets find the actual line that minimizes the RSS error term

```python
# Now lets find the actual line that minimizes the RSS
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x.reshape(-1, 1), y)

b = model.intercept_ # calculate intercept
m = model.coef_ # calculate slope

# Calculate y values for the line only at given x values
line_eq = f"y = {b} + {m}x"
color = "gold"
y_line = b + m * x

# Create the plot
plt.figure(figsize=(6, 4))

# Scatter plot of actual data points
plt.scatter(x, y, color='gold', label='Actual Data Points', marker='o')

# Scatter plot for y = 10x + 5 at given x values
plt.scatter(x, y_line, color=color, label=f'Line Points ({line_eq})', marker='x')

# Connect the red points to form a line
plt.plot(x, y_line, color=color, linestyle='--')

# Annotate each point
for i in range(len(x)):
    plt.text(x[i], y[i], f"({x[i]},{y[i]})", fontsize=10, verticalalignment='bottom', horizontalalignment='right')
    plt.text(x[i], y_line[i], f"({x[i]},{y_line[i]})", fontsize=10, verticalalignment='top', horizontalalignment='left', color=color)

# Set the x and y axis limits and ticks
plt.xlim(0, 6) # Set x-axis limits from 0 to 6
plt.ylim(0, 70) # Set y-axis limits from 0 to 70
plt.yticks(np.arange(0, 71, 10)) # Set y-axis ticks with a step of 10

# Labels, title, and legend
plt.xlabel("Hours ")
plt.ylabel("Scores")
plt.title("Scatter Plot with Regression Line Points")
plt.legend()
# plt.grid(True)

# Show the plot
plt.show()

# Print table of original and predicted values
print("Original Data Points:")
print("\n".join(f"({x[i]}, {y[i]})" for i in range(len(x))))

print(f"\nLine Points ({line_eq}):")
print("\n".join(f"({x[i]}, {y_line[i]})" for i in range(len(x))))

# Compute RSS error terms
errors = (y - y_line) ** 2
RSS = np.sum(errors)  # Residual Sum of Squares

# Print error table
print("\nTable of Errors:")
print(f"{'X':<5}{'Actual Y':<10}{'Predicted Y':<15}{'Error Term (Squared)':<20}")
print("-" * 50)
for i in range(len(x)):
    print(f"{x[i]:<5}{y[i]:<10}{y_line[i]:<15}{errors[i]:<20.2f}")

# Print total RSS
print(f"\nResidual Sum of Squares (RSS): {RSS:.2f}")
```
