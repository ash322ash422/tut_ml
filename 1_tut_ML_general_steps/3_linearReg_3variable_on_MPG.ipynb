# Linear Regression - Multiple Variable : MPG dataset

You are given a CSV file that contains automobile data: **MPG, cylinders, weight, model_year, etc**. Design ML model that can make prediction on **MPG** based on **cylinders, weight, model_year**.

Steps:

```
--> Collect data
--> Split into training + testing
--> Feed training data to model (i.e. train the model)
--> Evaluate the model by feeding testing data
```


## Collect Data

````python
import pandas as pd

df = pd.read_csv("data_MPG.csv")
print(df)
````

````python
# Select Features(X) & Target(y)

X = df[['cylinders', 'weight', 'model_year']] # 
y = df['mpg']
````

## Split into training + testing 

````python
# Train / Test Split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split( X, y, test_size=0.2)
````

````python
# Lets see how many records are in training and testing dataset

print("X_train shape:", X_train.shape)
print(f"X_test shape:", X_test.shape)
````

## Feed training data to the model (i.e. train the model)

````python
# Train the Multiple Linear Regression Model

from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_train, y_train)
````

````python
# Lets make a prediction on MPG of a car at idx=2, cylinders=8, weight=3436, model_year=70
# and check against the above MPG data=18

X_sample = [[8, 3436, 70],]
X_df = pd.DataFrame(X_sample, columns=X_train.columns)

prediction = model.predict(X_df)

print("The prediction is", prediction)
````

##  Evaluate the model by feeding testing data

````python
from sklearn.metrics import r2_score

y_pred = model.predict(X_test)

r2_value = r2_score(y_test, y_pred)

print("R-square Score:", r2_value) # close to 1 is good
````

## OPTIONAL

````python
import matplotlib.pyplot as plt

plt.scatter(y_test, y_pred)
plt.xlabel("Actual MPG")
plt.ylabel("Predicted MPG")
plt.title("Actual vs Predicted MPG")
plt.show()

# Ideally this should lie on diagonal: prediction = actual
````
