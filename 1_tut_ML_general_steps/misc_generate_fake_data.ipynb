# Generate a fake data that has 100 rows and 4 features A,B,C,D. The label y should be binary (0 or 1)

```python
from sklearn.datasets import make_multilabel_classification


X, y = make_multilabel_classification(n_classes=1, 
                                      n_labels=1,
                                      n_features=4,
                                      random_state=42)
```

```python
print(X[:5])
```

```python
print(y[:5])
```

```python
# value counts of y
import numpy as np

unique_values, counts = np.unique(y, return_counts=True)
print(unique_values)
print(counts)
```

```python
import pandas as pd

df =pd.DataFrame(X, columns=["A", "B", "C", "D"])
df['target'] = y

print(df)
```
