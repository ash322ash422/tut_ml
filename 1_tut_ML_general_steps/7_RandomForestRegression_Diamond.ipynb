# Random Forest Regression: 
You are given a CSV file that contains diamonds data and their **price**. 


Design ML model that can make prediction on **price** based on diamond data **'carat', 'depth', 'table', 'x', 'y', 'z'**.


Steps:

```
--> Collect data
--> Split into training + testing
--> Feed training data to model (i.e. train the model)
--> Evaluate the model by feeding testing data
```

## Collect data

````python
# Load dataset

import pandas as pd

import seaborn as sns
df = sns.load_dataset("diamonds")

# df = pd.read_csv("data_diamonds.csv")
print(df)
````

````python
# Seperate your data into feature X and response y
columns = ['carat', 'depth', 'table', 'x', 'y', 'z']

X = df[columns] # This is the feature that would predict the y
y = df["price"] # This is th prediction: labels/response/target
````

## Now we split the dataset into training / testing

````python
# Train-test split: 
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split( X, y, test_size=0.2 )
````

````python
# Lets see how many records are in training and testing dataset

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
````

## Feed training data to the model (i.e. train the model)

````python
# Took 1 minute: : Train Random Forest Regression
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor()

model.fit(X_train, y_train)
````

````python
# Print the data X at idx=2 and its label
print(X.columns.to_list())
print(X.iloc[2].values)
print(y.iloc[2])
````

````python
# Lets make a prediction on price for a diamond at idx=2 for 
# and check against actual price = 327
# ['carat', 'depth', 'table', 'x', 'y', 'z']
X_sample = [[ 0.23, 56.9, 65, 4.05, 4.07, 2.31],]

X_df = pd.DataFrame(X_sample, columns=X_train.columns)

prediction = model.predict(X_df)

print("The prediction is", prediction)
````

## Evaluate the model by feeding the testing data

````python
# step1: Lets make prediction on test data set
y_pred = model.predict(X_test)
````

````python
# lets print actual and predicted price

print("actual   :", y_test[:6].values)
print("predicted:", list(y_pred[:6]))
````

````python
# step3: Lets see the performance metric of the model 
from sklearn.metrics import r2_score

# Compute R-squared value
r2 = r2_score(y_test, y_pred)

print(f'R-squared: {r2:.4f}')  # closer to 1 is good
````
