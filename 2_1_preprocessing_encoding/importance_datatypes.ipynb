### Now lets save memory: Memory optimization

```python
# Lets print memory usage before memory optimization

print(df.memory_usage(deep=True))
print(f"Before optimization: {df.memory_usage(deep=True).sum() / 1024:.2f} KB")
```

```python
# 1) The hard way:
df['Age'] = df['Age'].astype('int8')

print(df.memory_usage(deep=True)['Age']) 
# memory consumption went down 11760 b -> 1470
```

```python
# 2) The  easy way: Let pandas decide based on range of values

# Downcast integer int64 columns to int8, int16, int32, int64 based on the data range present
cols = df.select_dtypes(include=['int64']).columns

df[cols] = df[cols].apply(pd.to_numeric, downcast='integer')
```

```python
print(df.memory_usage(deep=True))
print(f"Memory now: {df.memory_usage(deep=True).sum() / 1024:.2f} KB")

# observ: DailyRate memory consumption went down from 11760 B -> 2940 B
# DailyRate memory consumption went down from 11760 B -> 1470 B, etc
```

```python
# check if downcast worked

print(df.info(memory_usage='deep'))
```

```python
# Downcast integer float64 columns  based on the data range present

cols = df.select_dtypes(include=['float64']).columns
df[cols] = df[cols].apply(pd.to_numeric, downcast='float')
```

```python
print(df.memory_usage(deep=True))
print(f"Memory now: {df.memory_usage(deep=True).sum() / 1024:.2f} KB")
```

```python
print(df.memory_usage(deep=True))
print(f"Memory now: {df.memory_usage(deep=True).sum() / 1024:.2f} KB")

# observ: 
```

```python
# Convert object to Categorical Text Columns
obj_cols = df.select_dtypes(include=['object']).columns

for col in obj_cols:
    num_unique = df[col].nunique()
    num_total = len(df[col])
    if num_unique / num_total < 0.5:  # heuristic: if < 50% unique
        print(f"Converting {col} to datatype category")
        df[col] = df[col].astype('category')
```

```python
print(df.memory_usage(deep=True))
print(f"Memory now: {df.memory_usage(deep=True).sum() / 1024:.2f} KB")
```

```python
# check if downcast worked

print(df.info(memory_usage='deep'))
```

```python
# 0) Find binary columns the hard way: Try every columns
# ['Unnamed: 0', 'Age', 'Attrition', 'DailyRate', 'Department', 'DistanceFromHome', 
#  'Education', 'EducationField', 'EmployeeNumber', 'Gender', 'HourlyRate', 'JobLevel',
#  'JobSatisfaction', 'MaritalStatus', 'TravelTime', 'email']

columns = ["EducationField"]
# df[columns].value_counts()
df[columns].nunique()

# bin_columns are  = ['Attrition', 'Gender']
```

```python
# 1) Easy way: find binary columns

binary_cols = []

for col in df.columns:
    if df[col].dropna().nunique() == 2:
        binary_cols.append(col)

binary_cols
```

```python
# 2) find value counts
for c in binary_cols:
    print(df[c].value_counts(), "\n")
```

```python
# 3) convert th binary column to bool data type

df['Attrition'] = df['Attrition'].map({'Yes': True, 'No': False})
df['Attrition'] = df['Attrition'].astype('bool')

df['Gender'] = df['Gender'].map({'Male': True, 'Female': False})
df['Gender'] = df['Gender'].astype('bool')
```

```python
print(df.memory_usage(deep=True))
print(f"Memory now: {df.memory_usage(deep=True).sum() / 1024:.2f} KB")
```

```python
# check if downcast worked

print(df.info(memory_usage='deep'))
```
