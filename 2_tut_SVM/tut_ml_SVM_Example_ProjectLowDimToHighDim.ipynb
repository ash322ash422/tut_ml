# projection of 1D into 2D

```python
import numpy as np
import matplotlib.pyplot as plt

# 1D input
x = np.array([-2, -1, 0, 1, 2])
y = np.array([-1, -1, 1, -1, -1])  # labels

# 2D mapping: phi(x) = [x, x^2]
x_mapped = np.vstack([x, x**2]).T

# Plot in original 1D and mapped 2D (vertically stacked)
plt.figure(figsize=(7, 10))

# Plot original 1D space
plt.subplot(2, 1, 1)
plt.scatter(x, np.zeros_like(x), c=y, cmap='bwr', s=100)
plt.title("Original 1D Space")
plt.yticks([])
plt.xlabel("x")
plt.grid(True)

# Plot mapped 2D space
plt.subplot(2, 1, 2)
plt.scatter(x_mapped[:, 0], x_mapped[:, 1], c=y, cmap='bwr', s=100)
plt.title("Mapped 2D Space via φ(x) = [x, x²]")
plt.xlabel("x")
plt.ylabel("x²")
plt.grid(True)

plt.tight_layout()
plt.show()
```

# projection of 2D into 3D

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Generate 10 points around a circle and one at the center
theta = np.linspace(0, 2 * np.pi, 10, endpoint=False)
radius = 1

# Circle points (class +1)
x_circle = np.column_stack((radius * np.cos(theta), radius * np.sin(theta)))
y_circle = np.ones(10)

# Center point (class -1)
x_center = np.array([[0, 0]])
y_center = np.array([-1])

# Combine
X = np.vstack((x_circle, x_center))
y = np.concatenate((y_circle, y_center))

# Mapping function φ(x1, x2) = (x1, x2, x1² + x2²)
def phi(X):
    x1, x2 = X[:, 0], X[:, 1]
    z = x1**2 + x2**2
    return np.column_stack([x1, x2, z])

X_mapped = phi(X)

# Plot 2D view
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title("Original 2D Data")
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='bwr', s=100)
plt.xlabel('x1')
plt.ylabel('x2')
plt.axhline(0, color='gray', linestyle='--', linewidth=0.5)
plt.axvline(0, color='gray', linestyle='--', linewidth=0.5)
plt.grid(True)

# Plot 3D mapped view
fig = plt.figure(figsize=(10, 5))
ax = fig.add_subplot(122, projection='3d')
ax.set_title("Mapped 3D Space: φ(x1, x2) = (x1, x2, x1² + x2²)")
ax.scatter(X_mapped[:, 0], X_mapped[:, 1], X_mapped[:, 2], c=y, cmap='bwr', s=100)

ax.set_xlabel('x1')
ax.set_ylabel('x2')
ax.set_zlabel('x1² + x2²')

plt.tight_layout()
plt.show()
```
