```python
# Data setup: Column N is the label.

df_normal['N'] = 1   # 1 is normal
df_abnormal['N'] = 0 # 0 is abnormal

print(df_normal.head(3))
print(df_abnormal.head(3))
```

```python
df = pd.concat([df_normal,df_abnormal], ignore_index=True)
```

```python
# 
X = df.iloc[:,:188] # all rows, col from 0 to 187, except the last column because it is label reponse variable'N'
y = df['N'] # Column 'N' is the target variable 

# print(X.sample(4)) 
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2,random_state=2,shuffle=True)
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
# lets get the docs
KNeighborsClassifier?
```

```python
model_knn = KNeighborsClassifier()
```

```python
model_knn.fit(X_train,y_train)
```

```python
predict_knn = model_knn.predict(X_test)
# print(predict_knn)
```

```python
print(predict_knn.shape)
```

```python
print(predict_knn[5:10])
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix,classification_report
```

```python
print("Accuracy score of KNN Model: ", accuracy_score(predict_knn, y_test))
```

```python
print("Confusion Matrix:\n")
print(confusion_matrix(predict_knn, y_test))
```

```python
print("Classification Report: ")
print(classification_report(predict_knn, y_test))
```

```python
# Different values of n

nn = [1, 2, 5, 10, 15]

for n in nn:
    kn = KNeighborsClassifier(n)
    kn.fit(X_train, y_train)
    p_kn = kn.predict(X_test)
    print(f"Accuracy score for n = {n} is {accuracy_score(p_kn, y_test):0.4f}.")
    
```
